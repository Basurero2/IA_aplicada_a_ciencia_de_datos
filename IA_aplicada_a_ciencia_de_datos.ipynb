> #### <strong><font color="7b71ff" size=6>Prompt:</font></strong>

Tenemos un DataFrame llamado "df" y queremos construir un visual que presente cuáles métodos de pago fueron los más utilizados.

En "df", tenemos la columna "metodo_pago" con los 4 tipos de pago utilizados en la compra.

Inicialmente, queremos construir una tabla llamada "metodos_de_pago" que resuma los datos de esta columna y, a continuación, deseamos crear una visualización que muestre la cantidad de esos métodos de pago presentes en la tabla "metodo_pago".

Utiliza las bibliotecas de visualización de datos de Python para crear esta visualización.


````python
import matplotlib.pyplot as plt

# 1. Crear la tabla resumen 'metodos_de_pago'
# Contamos cuántas veces aparece cada método de pago en la columna 'metodo_pago' del DataFrame 'df'
metodos_de_pago = df['metodo_pago'].value_counts().reset_index()

# Renombramos las columnas para que sean claras
metodos_de_pago.columns = ['metodo_pago', 'cantidad']
metodos_de_pago
````

````python
# 2. Crear la visualización (Gráfico de Barras)
plt.figure(figsize=(10, 6))

# Usamos un gráfico de barras porque estamos comparando categorías
barras = plt.bar(metodos_de_pago['metodo_pago'], metodos_de_pago['cantidad'], color='#4c72b0')

# Añadimos etiquetas y título para que el gráfico se entienda por sí solo
plt.xlabel('Método de Pago')
plt.ylabel('Cantidad')
plt.title('Distribución de los Métodos de Pago en 2023')
plt.xticks(rotation=45) # Rotamos las etiquetas del eje X para que no se solapen

# Añadimos el valor exacto encima de cada barra para facilitar la lectura
for barra in barras:
    print(f"posisiones de las barras : {barra.get_x()}")
    altura = barra.get_height()
    plt.text(barra.get_x() + barra.get_width()/2., altura,
             f'{int(altura)}',
             ha='center', va='bottom')

# Ajustamos el diseño y mostramos el gráfico
plt.tight_layout()
plt.savefig('metodos_pago_distribucion.png') # Guardamos la imagen
plt.show()
````

### **2da Pregunta:** ¿Cuál fue la facturación por categoria em 2023?

**Observación:** Para la facturación, consideraremos aquí el producto entre el precio unitario y la cantidad, sumados al costo de envío de cada producto.


> #### <strong><font color="7b71ff" size=6>Prompt:</font></strong>

Tenemos un DataFrame llamado "df" y queremos construir un gráfico de barras horizontales que muestre la facturación por categoría de Zoop.

En "df", tenemos las siguientes columnas:
* categoria: categorías de los productos
* precio_unitario: precio de una unidad del producto
* cantidad: cantidad vendida del producto
* envio: costo de envío total de la compra.

Deseamos crear una visualización que muestre la facturación de cada categoría, donde la facturación se calcula multiplicando el precio unitario por la cantidad y sumando el costo de envío.

Utiliza las bibliotecas de visualización de datos de Python para crear esta visualización.


````python
# 1. Crear la columna de 'facturacion' según tu fórmula
# (Precio x Cantidad) + Envio
df['facturacion'] = (df['precio_unitario'] * df['cantidad']) + df['envio']

# 2. Agrupar por categoría y sumar la facturación
# Usamos .sort_values(ascending=True) para que al graficar, la categoría mayor quede arriba.
facturacion_por_categoria = df.groupby('categoria')['facturacion'].sum().sort_values(ascending=True)

# 3. Crear el gráfico de barras horizontales
plt.figure(figsize=(12, 6))

# 'barh' es la función clave para barras horizontales
barras = plt.barh(facturacion_por_categoria.index, facturacion_por_categoria.values, color='#2ca02c')

# 4. Personalización
plt.xlabel('Facturación Total MXM($)')
plt.ylabel('Categoría de Producto')
plt.title('Ranking de Facturación por Categoría en Zoop')

# 5. Añadir etiquetas de valor al final de cada barra
# Al ser horizontal, cambiamos la lógica: x es el ancho de la barra, y es la posición vertical.
for barra in barras:
    ancho = barra.get_width() # Esto es el valor monetario
    plt.text(
        ancho * 1.01, # Colocamos el texto un 1% a la derecha del final de la barra para dar aire
        barra.get_y() + barra.get_height()/2, # Centrado verticalmente en la barra
        f'${ancho:,.0f}', # Formato de moneda: $1,234 (sin decimales para limpieza)
        va='center', # Alineación vertical centrada
        ha='left'    # Alineación horizontal a la izquierda (el texto empieza donde termina la barra)
    )

# Ajuste de márgenes para que no se corte el texto largo
plt.tight_layout()
````

* facturacion: con los valores de las ventas
* metodo_pago: con el método de pago elegido en la compra

````python
# 2. Creamos la columna de 'Trimestre'
# .dt.to_period('Q') convierte la fecha a formato Trimestre (ej: 2023Q1)
df['trimestre'] = df['fecha'].dt.to_period('Q')

# 3. Agrupamos los datos
# Necesitamos sumar la facturación desglosada por Trimestre Y por Método de Pago
# .unstack() es el truco clave: convierte los métodos de pago de filas a columnas
ventas_trimestrales = df.groupby(['trimestre', 'metodo_pago'])['facturacion'].sum().unstack()

# 4. Crear el gráfico de barras apiladas
# kind='bar' + stacked=True crea el apilamiento
ax = ventas_trimestrales.plot(kind='bar', stacked=True, figsize=(10, 6), colormap='viridis')

# 5. Personalización
plt.title('Evolución Trimestral de Ventas por Método de Pago - Zoop 2023')
plt.xlabel('Trimestre')
plt.ylabel('Facturación Total MXM($)')
plt.xticks(rotation=45) # Mantiene las etiquetas horizontales (Q1, Q2...) para facilitar lectura

# Mover la leyenda fuera del gráfico para que no tape las barras
plt.legend(title='Método de Pago', bbox_to_anchor=(0, 1), loc='upper left')

# Formato de moneda en eje Y
current_values = plt.gca().get_yticks()
plt.gca().set_yticklabels(['${:,.0f}'.format(x) for x in current_values])

plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()

# Guardar y mostrar
plt.savefig('ventas_trimestrales_apiladas.png')
plt.show()
````

¿Por qué un gráfico de columnas agrupadas?

Un gráfico de columnas agrupadas es ideal para comparar categorías diferentes en relación con una variable (puede ser temporal o no), mostrando cómo los valores se comparan dentro de cada categoría en columnas separadas, facilitando la visualización de patrones y diferencias entre grupos.

````python
import matplotlib.pyplot as plt

# 1. Definir las categorías de interés
# Creamos una lista para filtrar solo lo que nos pediste
categorias_seleccionadas = ["Papelería y oficina", "Libros", "Alimentos y bebidas"]

# 2. Filtrar el DataFrame
# Nos quedamos solo con las filas que coincidan con esas 3 categorías
df_filtrado = df[df['categoria'].isin(categorias_seleccionadas)]

# 3. Agrupar y Reorganizar (Pivotar)
# Agrupamos por Trimestre y Categoría, sumamos la facturación.
# .unstack() es el comando mágico que mueve las categorías de las filas a las columnas,
# permitiendo que pandas las grafique una al lado de la otra automáticamente.
ventas_agrupadas = df_filtrado.groupby(['trimestre', 'categoria'])['facturacion'].sum().unstack()

# 4. Crear el gráfico
# kind='bar' por defecto agrupa lado a lado (si no ponemos stacked=True)
ax = ventas_agrupadas.plot(kind='bar', figsize=(12, 6), width=0.8, colormap='viridis')

# 5. Personalización
plt.title('Comparativa de Ventas por Trimestre: Libros vs. Papelería vs. Alimentos')
plt.xlabel('Trimestre')
plt.ylabel('Facturación ($)')
plt.xticks(rotation=0) # Mantiene Q1, Q2 horizontales
plt.legend(title='Categoría')
plt.grid(axis='y', linestyle='--', alpha=0.3)

# Formato de moneda en el eje Y (opcional para limpieza visual)
current_values = plt.gca().get_yticks()
plt.gca().set_yticklabels(['${:,.0f}'.format(x) for x in current_values])

plt.tight_layout()
plt.savefig('ventas_trimestrales_agrupadas.png')
plt.show()
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Paso 1: Preparar los datos
# Calculamos el conteo de valores en la columna 'cashback'
# Esto nos da la cantidad de usuarios que 'Sí' y 'No' participan
conteo_cashback = df['cashback'].value_counts()

# Paso 2: Configuración del gráfico
plt.figure(figsize=(8, 8))

# Paso 3: Crear el gráfico de pastel
# 'autopct' nos permite mostrar el porcentaje calculado automáticamente en cada rebanada
plt.pie(
    conteo_cashback,
    labels=conteo_cashback.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=sns.color_palette("pastel"), # Uso de paleta de colores estética
    explode=(0.05, 0), # 'explode' separa ligeramente la primera rebanada para darle énfasis
    shadow=True
)

plt.title('Proporción de Usuarios en el Programa Cashback (Zoop)')
plt.axis('equal') # Asegura que el gráfico sea un círculo perfecto

# Guardamos la imagen
plt.savefig('grafica_proporcion_cashback.png')
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# 1. Simulación del DataFrame 'df' (si no existe)
# Generamos 500 notas simuladas con una distribución realista (más notas altas que bajas)
# En tu caso, simplemente usarías tu df existente.
# np.random.seed(42) ... (código de simulación omitido para brevedad)

# 2. Configuración de la Visualización
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# Utilizamos sns.histplot para visualizar la distribución.
# 'discrete=True': Importante para tratar las notas como enteros (0, 1, 2...) y no como rangos continuos.
# 'kde=True': Añade una línea de Estimación de Densidad Kernel (la curva suave) para ver la tendencia.
ax = sns.histplot(
    data=df,
    x='nota',
    discrete=True,
    kde=False,
    color='#6c5ce7', # Color moderno (púrpura/azulado)
    alpha=0.7        # Transparencia para ver la cuadrícula de fondo
)

# 3. Personalización y Etiquetas
plt.title('Distribución de Evaluaciones de Compras (0-10)', fontsize=16)
plt.xlabel('Nota Asignada', fontsize=12)
plt.ylabel('Frecuencia (Cantidad de Usuarios)', fontsize=12)

# Asegurar que el eje X muestre todos los enteros del 0 al 10 claramente
plt.xticks(range(0, 11))
plt.xlim(-0.5, 10.5) # Márgenes para que las barras de 0 y 10 no se corten

# Añadir etiquetas de valor sobre cada barra
for p in ax.patches:
    height = p.get_height()
    if height > 0: # Solo etiquetar si hay datos
        ax.text(p.get_x() + p.get_width() / 2., height + 1, int(height), ha="center", fontsize=10)

plt.tight_layout()
plt.savefig('distribucion_notas_compra.png')
````
